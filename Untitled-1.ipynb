```python
# import libs
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from backtesting import Backtest, Strategy
from datetime import time
```

```python
# Read CSV minute data
working_data = pd.read_csv('data.csv', parse_dates=['Date'], low_memory=False)
working_data
```

```python
df = working_data.copy()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
df.index = pd.DatetimeIndex(df.index)
 
# backtesting.py requires: Open, High, Low, Close, Volume
df = df.rename(columns={"tick_volume": "Volume"})
df = df[["Open", "High", "Low", "Close", "Volume"]]
df = df.sort_index()
```

```python
# ──────────────────────────────────────────────
# 2. PRECOMPUTE DAILY RTH STATS
#    RTH = 09:30 – 16:00 NY time
# ──────────────────────────────────────────────
rth_bars = df.between_time("09:30", "16:00")
 
daily_stats = (
    rth_bars.groupby(rth_bars.index.date)
    .agg(rth_high=("High", "max"), rth_low=("Low", "min"))
)
# Store as a dict keyed by date for fast lookup
daily_stats_dict = daily_stats.to_dict("index")
trading_dates = sorted(daily_stats.index.tolist())
```

```python
class RTHGapFill(Strategy):
    min_gap = 5.0

    def init(self):
        self.traded_today = None

    def next(self):
        current_dt   = self.data.index[-1]
        current_date = current_dt.date()
        current_hour = current_dt.hour
        current_min  = current_dt.minute

        # ── CLOSE ALL POSITIONS at 15:59 before RTH close ──
        if current_hour == 15 and current_min == 59:
            if self.position:
                self.position.close()
            return

        # ── Only enter at 9:30 AM ──
        if not (current_hour == 9 and current_min == 30):
            return

        if self.traded_today == current_date:
            return

        if current_date not in daily_stats_dict:
            return

        date_idx = trading_dates.index(current_date) if current_date in trading_dates else -1
        if date_idx <= 0:
            return

        prev_date = trading_dates[date_idx - 1]
        if prev_date not in daily_stats_dict:
            return

        prev_high  = daily_stats_dict[prev_date]["rth_high"]
        prev_low   = daily_stats_dict[prev_date]["rth_low"]
        open_price = self.data.Open[-1]

        # ── GAP UP → SHORT ──
        if open_price - prev_high > self.min_gap:
            gap_size = open_price - prev_high
            tp = prev_high
            sl = open_price + gap_size
            self.sell(limit=open_price, sl=sl, tp=tp)
            self.traded_today = current_date

        # ── GAP DOWN → LONG ──
        elif prev_low - open_price > self.min_gap:
            gap_size = prev_low - open_price
            tp = prev_low
            sl = open_price - gap_size
            self.buy(limit=open_price, sl=sl, tp=tp)
            self.traded_today = current_date
```

```python
# ──────────────────────────────────────────────
# 4. RUN BACKTEST
# ──────────────────────────────────────────────
bt = Backtest(
    df,
    RTHGapFill,
    cash=100_000,
    commission=0.0,         # adjust if needed (e.g. 0.0002)
    exclusive_orders=True,  # cancel pending orders on new signal
)
 
stats = bt.run()
```

```python
# Show back-testing results
stats
```

```python
#  Copy all trades data to trades df
trades = stats['_trades'].copy()
# Doing simlpe stats on the trades data
trades['Direction'] = trades['Size'].apply(lambda x: 'Long' if x > 0 else 'Short')
trades['EntryYear'] = pd.to_datetime(trades['EntryTime']).dt.year
trades['EntryDay']  = pd.to_datetime(trades['EntryTime']).dt.day_name()
trades['IsWin']     = trades['PnL'] > 0

# Long vs Short
print("=== LONG vs SHORT ===")
print(trades.groupby('Direction').agg(
    Trades  = ('PnL', 'count'),
    WinRate = ('IsWin', lambda x: f"{x.mean()*100:.1f}%"),
    TotalPnL= ('PnL', 'sum'),
    AvgPnL  = ('PnL', 'mean'),
).to_string())

# By year
print("\n=== BY YEAR ===")
print(trades.groupby('EntryYear').agg(
    Trades  = ('PnL', 'count'),
    WinRate = ('IsWin', lambda x: f"{x.mean()*100:.1f}%"),
    TotalPnL= ('PnL', 'sum'),
).to_string())

# By day
print("\n=== BY DAY ===")
day_order = ['Monday','Tuesday','Wednesday','Thursday','Friday']
print(trades.groupby('EntryDay').agg(
    Trades  = ('PnL', 'count'),
    WinRate = ('IsWin', lambda x: f"{x.mean()*100:.1f}%"),
    TotalPnL= ('PnL', 'sum'),
).reindex(day_order).to_string())

# Exit reason breakdown
print("\n=== EXIT REASON (how trades closed) ===")
print(f"Hit TP  : {(trades['ExitPrice'] == trades['TP']).sum()}")
print(f"Hit SL  : {(trades['ExitPrice'] == trades['SL']).sum()}")
print(f"EOD     : {((trades['ExitPrice'] != trades['TP']) & (trades['ExitPrice'] != trades['SL'])).sum()}")
```

```python
# Filter only trades that hit stop loss or take profit
trades = trades[(trades['ExitPrice'] == trades['TP']) | (trades['ExitPrice'] == trades['SL'])]
# Show resualt dataframe
trades
```

```python

# Filter Winning Trades
winning = trades[trades['PnL'] > 0][[
    'EntryTime', 'ExitTime', 'Size', 'EntryPrice', 'ExitPrice', 'PnL', 'ReturnPct', 'Duration'
]].copy()

# Add direction
winning['Direction'] = winning['Size'].apply(lambda x: 'Long' if x > 0 else 'Short')

# Convert times to NY timezone
winning['EntryTime_NY'] = winning['EntryTime'].dt.tz_convert('America/New_York')
winning['ExitTime_NY']  = winning['ExitTime'].dt.tz_convert('America/New_York')

# Clean display
winning_trades = winning[[
    'EntryTime_NY', 'ExitTime_NY', 'Direction',
    'EntryPrice', 'ExitPrice', 'PnL', 'ReturnPct', 'Duration'
]].reset_index(drop=True)

winning_trades
```

```python
# Sperate winning long trades from winning short trades
winning_trades_long = winning_trades[winning_trades['Direction'] == 'Long']
winning_trades_short = winning_trades[winning_trades['Direction'] == 'Short']
```

```python
# Filter lossing Trades
losses = trades[trades['PnL'] < 0][[
    'EntryTime', 'ExitTime', 'Size', 'EntryPrice', 'ExitPrice', 'PnL', 'ReturnPct', 'Duration'
]].copy()

# Add direction
losses['Direction'] = losses['Size'].apply(lambda x: 'Long' if x > 0 else 'Short')

# Convert times to NY timezone
losses['EntryTime_NY'] = losses['EntryTime'].dt.tz_convert('America/New_York')
losses['ExitTime_NY']  = losses['ExitTime'].dt.tz_convert('America/New_York')

# Clean display
losses_trades = losses[[
    'EntryTime_NY', 'ExitTime_NY', 'Direction',
    'EntryPrice', 'ExitPrice', 'PnL', 'ReturnPct', 'Duration'
]].reset_index(drop=True)

losses_trades
```

```python
# Sperate lossing long trades from lossing short trades
losses_trades_long = losses_trades[losses_trades['Direction'] == 'Long']
losses_trades_short = losses_trades[losses_trades['Direction'] == 'Short']
```

```python
# Copy 1 mins timeframe data into new df
minute_dff = working_data.copy()
# Ensure 'Date' column is in datetime format
minute_dff['Date'] = pd.to_datetime(minute_dff['Date'])

# Set 'Date' as the index
minute_dff.set_index('Date', inplace=True)

# Verify the data
minute_dff.head()
```

```python
# Read CSV daily volume profile RTH data
volume_profile_RTH = pd.read_csv('daily_profile_RTH.csv', parse_dates=['date'], low_memory=False)
volume_profile_RTH
```

```python
# Read CSV daily volume profile PreMarket data
volume_profile_premarket = pd.read_csv('daily_profile_premarket.csv', parse_dates=['date'], low_memory=False)
volume_profile_premarket
```

```python
# %% 
# FIX: Create correct NY session date FIRST
minute_dff['date'] = (
    pd.to_datetime(minute_dff.index)
    .tz_convert('America/New_York')   # align to trading session
    .tz_localize(None)                # remove timezone for merge
    .normalize()
)

# %% 
# FIX: Clean premarket dates
volume_profile_premarket['date'] = pd.to_datetime(
    volume_profile_premarket['date'].astype(str),
    errors='coerce'
).dt.normalize()

# Set index for join
volume_profile_premarket = volume_profile_premarket.set_index('date')

# %% 
# FIX: Merge (replace old join)
minute_dff = minute_dff.merge(
    volume_profile_premarket,
    left_on='date',
    right_index=True,
    how='left'
)
```

```python
# Get daily VP values and shift by 1 day to get "yesterday's" values
daily_vp = volume_profile_RTH.groupby('date')[['high', 'low', 'poc', 'val', 'vah']].first()
daily_vp_yesterday = daily_vp.shift(1).rename(columns={
    'high': 'yest_high',
    'low':  'yest_low',
    'poc':  'yest_poc',
    'val':  'yest_val',
    'vah':  'yest_vah'
})

minute_dff = minute_dff.join(daily_vp_yesterday, on='date', how='left')
minute_dff
```

```python
# Now let's get daily data for just the long wins
winning_dates_long = set(winning_trades_long['EntryTime_NY'].dt.date)
winning_data_long = minute_dff[pd.Series(minute_dff.index.date, index=minute_dff.index).isin(winning_dates_long)]
wining_data_long = winning_data_long[winning_data_long.index.time == time(9, 31)]

# Now let's get daily data for just the short wins
winning_dates_short = set(winning_trades_short['EntryTime_NY'].dt.date)
winning_data_short = minute_dff[pd.Series(minute_dff.index.date, index=minute_dff.index).isin(winning_dates_short)]
wining_data_short = winning_data_short[winning_data_short.index.time == time(9, 31)]
```

```python
# Now let's get daily data for just the long losses
lossing_dates_long = set(losses_trades_long['EntryTime_NY'].dt.date)
lossing_data_long = minute_dff[pd.Series(minute_dff.index.date, index=minute_dff.index).isin(lossing_dates_long)]
lossing_data_long = lossing_data_long[lossing_data_long.index.time == time(9, 31)]

# Now let's get daily data for just the short losses
lossing_dates_short = set(losses_trades_short['EntryTime_NY'].dt.date)
lossing_data_short = minute_dff[pd.Series(minute_dff.index.date, index=minute_dff.index).isin(lossing_dates_short)]
lossing_data_short = lossing_data_short[lossing_data_short.index.time == time(9, 31)]
```
